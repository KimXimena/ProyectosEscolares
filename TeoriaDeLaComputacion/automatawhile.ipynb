```python
estado = 0

cadena = input("Ingresa una palabra: ")

for letra in cadena:

    if estado == 0:
        if letra == 'w':
            estado = 1
        else:
            estado = -1

    elif estado == 1:
        if letra == 'h':
            estado = 2
        else:
            estado = -1

    elif estado == 2:
        if letra == 'i':
            estado = 3
        else:
            estado = -1

    elif estado == 3:
        if letra == 'l':
            estado = 4
        else:
            estado = -1

    elif estado == 4:
        if letra == 'e':
            estado = 5
        else:
            estado = -1

if estado == 5:
    print("Cadena aceptada")
else:
    print("Cadena rechazada")
```

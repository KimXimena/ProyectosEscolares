```python
def es_variable(token):
    if len(token) == 0:
        return False

    if not token[0].isalpha():
        return False

    for caracter in token:
        if not caracter.isalnum():
            return False

    return True


def es_numero(token):
    if len(token) == 0:
        return False

    for caracter in token:
        if not caracter.isdigit():
            return False

    return True


def es_valor(token):
    return es_variable(token) or es_numero(token)


def tokenizar(cadena):
    tokens = []
    i = 0

    while i < len(cadena):
        caracter = cadena[i]

        if caracter == " ":
            i += 1

        elif caracter.isalpha():
            palabra = ""

            while i < len(cadena) and cadena[i].isalnum():
                palabra += cadena[i]
                i += 1

            tokens.append(palabra)

        elif caracter.isdigit():
            numero = ""

            while i < len(cadena) and cadena[i].isdigit():
                numero += cadena[i]
                i += 1

            tokens.append(numero)

        elif caracter in ["(", ")"]:
            tokens.append(caracter)
            i += 1

        elif caracter in ["<", ">", "=", "!"]:
            operador = caracter

            if i + 1 < len(cadena) and cadena[i + 1] == "=":
                operador += "="
                i += 2
            else:
                i += 1

            tokens.append(operador)

        else:
            tokens.append(caracter)
            i += 1

    return tokens


def revisar_sintaxis(cadena):
    tokens = tokenizar(cadena)

    estado = "q0"

    for token in tokens:

        if estado == "q0":
            if token == "if" or token == "while":
                estado = "q3"
            elif es_variable(token):
                estado = "q1"
            else:
                estado = "qe"

        elif estado == "q1":
            if token == "=":
                estado = "q2"
            else:
                estado = "qe"

        elif estado == "q2":
            if es_numero(token):
                estado = "qf"
            else:
                estado = "qe"

        elif estado == "q3":
            if token == "(":
                estado = "q4"
            else:
                estado = "qe"

        elif estado == "q4":
            if es_valor(token):
                estado = "q5"
            else:
                estado = "qe"

        elif estado == "q5":
            if token in ["<", ">", "<=", ">=", "==", "!="]:
                estado = "q6"
            else:
                estado = "qe"

        elif estado == "q6":
            if es_valor(token):
                estado = "q7"
            else:
                estado = "qe"

        elif estado == "q7":
            if token == ")":
                estado = "qf"
            else:
                estado = "qe"

        elif estado == "qf":
            estado = "qe"

        if estado == "qe":
            break

    if estado == "qf":
        return True
    else:
        return False


texto = input("Ingresa una instrucción: ")

if revisar_sintaxis(texto):
    print("Sintaxis correcta. Cadena aceptada.")
else:
    print("Sintaxis incorrecta. Cadena rechazada.")
```

```python
from fpdf import FPDF

# Crear objeto PDF
pdf = FPDF()

# Agregar una página
pdf.add_page()

# Establecer fuente
pdf.set_font("Arial", size=12)

# Agregar texto
pdf.cell(200, 10, txt="Reporte de estudiantes", ln=True, align="C")

pdf.cell(200, 10, txt="Nombre: Juan Perez", ln=True)

pdf.cell(200, 10, txt="Promedio: 9.2", ln=True)

# Guardar el archivo
pdf.output("reporte1.pdf")
```

```python
#En este ejemplo utilizamos la librería FPDF en Python para generar un documento PDF sencillo.
#Primero se crea el objeto del documento, luego se añade una página, se define la fuente, se agregan
#las celdas con texto y finalmente se genera el archivo PDF.
```

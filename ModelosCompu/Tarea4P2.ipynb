```python
import numpy as np

pBB = 0.8
pNB = 0.3
P = np.array([[pBB, 1 - pBB], #estados en filas: B, N salen, columnas: B, N entran
              [pNB, 1 - pNB]], dtype=float)

population = 1000
buyers0 = 100
x0 = np.array([buyers0 / population, 1 - (buyers0 / population)], dtype=float) #B, N

x1 = x0 @ P#después de 1 mes
x2 = x0 @ np.linalg.matrix_power(P, 2)#después de 2 meses

buyers_1 = x1[0] * population
buyers_2 = x2[0] * population

print("Matriz de transición:\n", P)
print("Distribución 1 mes:", x1, "Compradores:", buyers_1)
print("Distribución 2 meses:", x2, "Compradores:", buyers_2)
```

```python
import numpy as np

def steady_state(P: np.ndarray):
    n = P.shape[0]
    A = P.T - np.eye(n)
    A[-1, :] = np.ones(n)#se sustituye la última ecuación por sum π=1
    b = np.zeros(n); b[-1] = 1.0
    return np.linalg.solve(A, b)

P = np.array([
    [0.85, 0.10, 0.05],#desde G
    [0.12, 0.80, 0.08],#desde H
    [0.15, 0.15, 0.70],#desde I
], dtype=float)

pi = steady_state(P)
print("Vector estacionario:", np.round(pi, 6))
```

```python
from math import factorial

lam = 8.0 
mu  = 11.0
s   = 3

a = lam / mu #tráfico ofrecido
rho = lam / (s * mu) #utilización

summ = sum(a**n / factorial(n) for n in range(s))
last = (a**s) / (factorial(s) * (1 - rho))
P0 = 1.0 / (summ + last)

Pw = last * P0

Lq = Pw * (lam * mu) * ((lam / mu) ** (s - 1)) / (factorial(s) * ((s * mu - lam) ** 2))

Wq = Lq / lam #tiempo de espera en cola (horas)
W  = Wq + 1.0 / mu #tiempo total en sistema (horas)
L  = lam * W #número de clientes en el sistema

print(f"rho (utilización): {rho:.4f}")
print(f"P0 (sistema vacío): {P0:.4f}")
print(f"Pw (prob. de esperar): {Pw:.4f}")
print(f"Lq (clientes en cola): {Lq:.4f}")
print(f"Wq (espera en cola, h): {Wq:.4f}")
print(f"W  (tiempo total, h): {W:.4f}")
print(f"L  (clientes en sistema): {L:.4f}")
```

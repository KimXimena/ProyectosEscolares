```python
import random
import numpy as np
import matplotlib.pyplot as pl
import math
```

```python
#configuracion para mostrar garficos en la vantana
pl.rcParams['figure.figsize'] = [10, 6]

print("Grafico de dispersion y su raiz cuadrada: \n")

#generar los 20 numeros aleatorios y calcular sus raices cuadradas
x_values=[random.randint(0,100) for _ in range(20)]
y_values=[math.sqrt(x) for x in x_values ]

#crear grafica
pl.figure()
pl.scatter(x_values, y_values, color='green', alpha=0.7)
pl.title('Gráfico de dispersión: Números aleatorios y su raíz cuadrada')
pl.xlabel('Números aleatorios (0-100)')
pl.ylabel('Raíz cuadrada')
pl.grid(True, linestyle='--', alpha=0.7)
pl.show()
```

```python
print("Producto esclar de dos matrices:\n")

#crear dos matrices de 3x3, con valores de -1000 a 1001
matrix_a=np.random.randint(-1000, 1001, size=(3,3))
matrix_b=np.random.randint(-1000, 1001, size=(3,3))

#calculamos el producto escalar
dot_product=np.dot(matrix_a,matrix_b)

#imprimimos
print("\n",matrix_a,"\n")
print("\n",matrix_b,"\n")
print(dot_product)
```

```python
print("Producto escalar de dos matrices con diferentes dimensiones: \n")

matrix_c=np.random.randint(-500, 501, size=(3,2))
matrix_d=np.random.randint(-500, 501, size=(2,2))
dott_product=np.dot(matrix_c,matrix_d)

print("\n",matrix_c,"\n")
print("\n",matrix_d,"\n")
print(dott_product)
```

```python
print("Producto de escalar de matrices de 3x3 y vector de 3 posiciones")

matrix_e=np.random.randint(-100, 101, size=(3,3))
matrix_f=np.random.randint(-100, 101, size=3)
dottt_product=np.dot(matrix_e, matrix_f)

print("\n",matrix_e,"\n")
print("\n",matrix_f,"\n")
print(dottt_product)
```

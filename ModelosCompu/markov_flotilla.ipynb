```python
import numpy as np
m1=np.matrix(([0.97,0.02,0.01],[0.6,0.3,0.1],[0.2,0.1,0.7]))
m1
```

```python
m_aux=np.matrix([(),(),()])
m_aux_anterior=np.matrix([(),(),()])
```

```python
for i in range(200):
    if i==0:
        m_aux=np.dot(m1,m1)
        print(m_aux)
        print("1 ------ ",+i+2)
    else:
        y=i+1
        m_aux_anterior=m_aux
        m_aux=np.dot(m_aux,m1)
        #if np.all(m_aux==m_aux_anterior):
        if np.array_equal(m_aux,m_aux_anterior): 
            print("Matriz de estado estable enontrada")
            print("*** numero de matri estable: "+str(y)+" ***")
            print(np.round(m_aux,3))
            print("N ------", y+1)
            break
        
        print(m_aux)
        print("N ------", y+1)
```

```python
#determinar el valor de la poblacion en el estado estable
poblacion_inicial=np.matrix([[100,0,0]])
```

```python
print("Comportamiento de la Poblacion")
for z in range(100): 
    if z==0:
        print("poblacion_inicial v0", poblacion_inicial)
        poblacion_inicial=np.dot(poblacion_inicial,m1)
        poblacion_inicial=np.round(poblacion_inicial,2)
        print(f"poblacion iteracion  v{z+1}",poblacion_inicial)
    else: 
        poblacion_anterior=poblacion_inicial
        poblacion_inicial=np.dot(poblacion_inicial,m1)
        poblacion_inicial=np.round(poblacion_inicial,2)
        if np.all(poblacion_inicial==poblacion_anterior):
            print("Poblacion en estado estable encontrada")
            print(f"*** numero de iteraciones: {z+1} ***")
            poblacion_inicial=np.round(poblacion_inicial,0)
            poblacion_inicial=poblacion_inicial.astype(int)
            print(np.round(poblacion_inicial,0))
            break
```

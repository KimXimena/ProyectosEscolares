```python
#PLANTEAMIENTO:
#UNA DULCERIA DE CINE TIENE TRES EMPLEADOS QUE ATIENDEN A LOS CLIENTES POR RIGUROSO ORDEN DE LLEGADA. EL TIEMPO DE SERVICIO POR CLIENTE SE DISTRIBUYE EXPONENCIALMENTE , CON UN PROMEDIO DE 2 MINUTOS POR CLIENTE. LOS CLIENTES DE LA DULCERIA ESPERAN EN UN AMPLIO ESPACIO FORMANDO UNA SOLA FILA Y SUS LLEGADAS TIENEN UNA DISTRIBUCION DE POISSON CON UN PROMEDIO DE 81 CLIENTES POR HORA. 
# SI EL TIEMPO PROMEDIO QUE PASAN EN EL AREA DE DULCERIA REBASA LOS 10 MINUTOS LOS CLIENTES SE SIENTEN INSATISFECHOS.
```

```python
#Servidores = 3
#Ritmo de servicio = 2 min por cliente = 30 clientes por hora
#Tasa de llegadas = 81 clientes por hora

#Conversion 2 minutos a horas
ritmo_servicio= 1/(2/60) #clientes por hora
print("Tasa de servicio por hora: ",ritmo_servicio)

 # 1/mu = 1/2 = 0.5 horas * 60 = 30 clientes por hora
```

```python
from math import exp, factorial
```

```python
#utilizacion promedio del sistema
def UtilizacionPromedio(tasadellegada,tasadeservicio,servidores):
    return float(tasadellegada / (servidores * tasadeservicio))
```

```python
def Pzero(tasadellegada,tasadeservicio,servidores):
    calculo_p1=0
    for x in range(servidores):
        calculo_p1 += ((tasadellegada/tasadeservicio)**x) / factorial(x)


    seccion1=((tasadellegada/tasadeservicio)**servidores) / factorial(servidores)
    seccion2=1/(1-UtilizacionPromedio(tasadellegada,tasadeservicio,servidores))

    result_seccion2=seccion1 * seccion2
    return 1 / (calculo_p1 + result_seccion2)

print(f"Pezero:", round(Pzero(81, ritmo_servicio, 3),4))
```

```python
#factor o porcentaje ocioso del sistema 
def Factor_O(tasadellegada, tasadeservicio, servidores):
    return (1 - UtilizacionPromedio(tasadellegada,tasadeservicio,servidores)) * 100
```

```python
print("Factor o porcentaje ocioso del sistema: ", round(Factor_O(81,ritmo_servicio,3),2),"%")
```

```python
#numero de clientes en cola Lq
def Numero_CC(tasadellegada, tasadeservicio, servidores):
    rho= UtilizacionPromedio(tasadellegada,tasadeservicio,servidores)
    P0=Pzero(tasadellegada,tasadeservicio,servidores)
    numerador = P0 *((tasadellegada/tasadeservicio)**servidores) * rho
    denominador = factorial(servidores) * ((1 -rho)**2)
    return numerador / denominador
```

```python
print("numero de clientes en cola Lq: ", round (Numero_CC(81,ritmo_servicio,3),4))
```

```python
#Tiempo promedio en cola
def Tiempo_PC (tasadellegada,tasadeservicio,servidores):
    Lq = Numero_CC(tasadellegada,tasadeservicio,servidores)
    return Lq / tasadellegada #horas
```

```python
print("tiempo promedio en cola wq: ", round(Tiempo_PC(81,ritmo_servicio,3)*60,4),"minutos")
```

```python
#tiempo promedio de permanecia en el sistema  W
def TiempoPP(tasadellegada,tasadeservicio,servidores):
    return Tiempo_PC(tasadellegada,tasadeservicio,servidores) + (1 / tasadeservicio)
```

```python
print("tiempo promedio en cola wq: ", round(TiempoPP(81,ritmo_servicio,3)*60,4),"minutos")
```

```python
#cantidad promedio de clientes en el sistema L
def CantidadPromedioClientesSistema_l(tasadellegada,tasadeservicio,servidores):
    return Numero_CC(tasadellegada,tasadeservicio,servidores) + (tasadellegada / tasadeservicio)
```

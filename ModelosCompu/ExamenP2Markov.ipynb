```python
import numpy as np
```

```python
#matriz
P = np.array([[0.5,  0.3,  0.2],[0.2, 0.5,  0.3],[0.15, 0.45, 0.4]], dtype=float)
P
```

```python
#estado inicial
inicio_dia = np.array([0.0, 1.0, 0.0])
def distribucion_dia(iniciar, P, dias):
    dists = [iniciar]
    i = iniciar.copy()
    for _ in range(dias):
        i = i @ P
        dists.append(i)
    return dists

inicio_dia
```

```python
dists = distribucion_dia(inicio_dia, P, 4)
lunes, martes, miercoles, jueves = dists[1], dists[2], dists[3], dists[4]
def pct(x): return f"{100*x:.4f}%"
```

```python
dias = ["Domingo", "Lunes", "Martes", "Miércoles", "Jueves"]
for i, d in enumerate(dists):
    print(f"{dias[i]:>14}: Soleado={pct(d[0])}  Nublado={pct(d[1])}  Lluvioso={pct(d[2])}")
```

```python
print(f"Probabilidad de que el martes sea soleado = {pct(martes[0])}")
print(f"Probabilidad de que el jueves sea lluvioso = {pct(jueves[2])}")
print(f"El jueves la probabilidad más alta (QUE DICE EN LA CELDA DE ARRUBA) es que esté nublado ({pct(jueves[1])}),")
print(f"mientras que la probabilidad de lluvia(QUE TAMBIEN MUESTRA LA CELDA DE ARRIBA) es menor ({pct(jueves[2])}).")
print("Por eso lo más probable es el jueves NO LLUEVA")
```

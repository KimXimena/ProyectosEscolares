```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#Datos
ventas = [120, 80, 150, 200, 90, 170, 130]
dias = ["Lun", "Mar", "Mié", "Jue", "Vie", "Sáb", "Dom"]

# Crear gráfica
plt.bar(dias, ventas, color='orange')
plt.title("Ventas Semanales de la Tienda")

#Etiquetas
plt.xlabel("Días de la Semana")
plt.ylabel("Número de Ventas")

#Promedio de ventas
p = np.mean(ventas)
plt.axhline(p, color='red', linestyle='--', label=f'Promedio: {p:.4f}')

#Mostrar valores arriba
for i, v in enumerate(ventas):
     plt.text(i, v + 1, str(v), ha='center', fontsize=10)

plt.legend(loc='upper left', bbox_to_anchor=(0.0, -0.1), fontsize=7)
plt.show()
```

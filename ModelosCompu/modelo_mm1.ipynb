```python
#PLANTEAMIENTO
#La gerente de una tienda de abarrotes, esta interesada en brindar un buen servicio a las personas mayores que compran en su tienda.
#Actualmente, la tienda tiene una
#caja registradora reservada para los clientes de la tercera edad. Esas personas llegan a la caja a un ritmo promedio de 30 
#clientes por hora, de acuerdo con una distribución de Poisson, y son atendidos
#a una tasa promedio de 35 clientes por hora, con tiempos de servicios exponenciales.
```

```python
#tasa de llegadas (LAMBDA)
LAMBDA = 30 #clientes por hora
#TASA DE SERVICIO (MU)
MU = 35  #clientes por hora
```

```python
vlamba=int(input("Ingrese la tasa de llegadas (clientes por hora): "))
vmu=int(input("Ingrese la tasa de servicio(clientes por hora):"))
```

```python
def cero_clientes(tasa_llegadas, ritmo_servicio):
    return(1 - (tasa_llegadas / ritmo_servicio)) * 100
```

```python
def utiliza_prom_sistema(tasa_llegadas, ritmo_servicio):
    return(tasa_llegadas / ritmo_servicio) * 100
```

```python
def promedio_clientes_sistema(tasa_llegadas, ritmo_servicio):
    return tasa_llegadas / (ritmo_servicio - tasa_llegadas)
```

```python
def promedio_clientes_cola_lq(tasa_llegadas,ritmo_servicio):
    return (tasa_llegadas**2) / (ritmo_servicio * (ritmo_servicio - tasa_llegadas))
```

```python
def promedio_permanencia_clientes_sistema(tasa_llegadas,ritmo_servicio):
    return 1  / (ritmo_servicio - tasa_llegadas)
```

```python
def promedio_espera_en_fila(tasa_llegadas):
    return promedio_clientes_cola_lq(vlamba,vmu) / tasa_llegadas
```

```python
#calcule la probabilidad de que no haya clientes en el sistema
print("Probabilidad de que no haya clientes en el sistema : ", cero_clientes(vlamba,vmu),"%")
```

```python
#utilizqacion promedio del sistema
print("Utilizacion promedio del sistema: ", utiliza_prom_sistema(vlamba,vmu),"%")
```

```python
#probabilidad promedio de clientes en el sistema
print("Promedio clientes en el sistema: ", promedio_clientes_sistema(vlamba,vmu), "%")
```

```python
#probabilidad de clientes en cola lq
print("promedio de clientes en la cola Lq: ", round(promedio_clientes_cola_lq(vlamba,vmu),0),"%")
```

```python
#probabilidad promedio de clientes en el sistema W
print("Promedio de permanecia de clientes en el sistema W: ", round(promedio_permanencia_clientes_sistema(vlamba,vmu),2),"horas")
print("Promedio de permanecia de clientes en el sistema W: ", round(promedio_permanencia_clientes_sistema(vlamba,vmu)*60, 2),"minutos")
```

```python
#probabilidad promedio de espera en la fila wq
print("Promedio de espera en la fila Wq: ",round(promedio_espera_en_fila(vlamba),2),"horas")
print("Promedio de espera en la fila Wq: ",round(promedio_espera_en_fila(vlamba)*60,2),"minutos")
```

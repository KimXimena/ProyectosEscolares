```python
from math import factorial
```

```python
lmbda = 25.0 #llegadas pr hora
lmbda
```

```python
mu = 7.0 #servidor por hora
mu
```

```python
s = 4 #numero de servidores
s
```

```python
#utilización
rho = lmbda / (s * mu)
if rho >= 1:
    raise ValueError("El sistema no es estable")
```

```python
#P0
a = lmbda/mu
suma_terms = sum((a**n) / factorial(n) for n in range(s))
p0_den = suma_terms + (a**s) / (factorial(s) * (1-rho))
P0=1/p0_den
```

```python
#Lq
Lq = P0 * (a**s) * rho / (factorial(s) * (1 - rho)**2)
```

```python
L = Lq + lmbda / mu
Wq = Lq / lmbda
W  = Wq + 1.0/mu
```

```python
print("Colas mms")
print(f"p = {rho:.4f}, P0 = {P0:.4f}")
print(f"Lq = {Lq:.4f} clientes")
print(f"L  = {L:.4f} clientes")
print(f"Wq = {Wq:.4f} h  = {Wq*60:.2f} min")
print(f"W  = {W:.4f} h  = {W*60:.2f} min")
```

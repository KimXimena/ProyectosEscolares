```python
import numpy as np
```

```python
P = np.array([
    [0.25, 0.50, 0.25],
    [0.05, 0.90, 0.05],
    [0.10, 0.50, 0.40]
])
```

```python
#vector de estado inicial
estado = np.array([1/3, 1/3, 1/3])
```

```python
#parámetros de convergencia
tolerancia = 1e-6
max_iter = 1000

for i in range(max_iter):
    nuevo_estado = np.dot(estado, P)
    
    #si la diferencia es menor que la tolerancia, se detiene
    if np.allclose(nuevo_estado, estado, atol=tolerancia):
        break
    
    estado = nuevo_estado
```

```python
print("Estado estable aprox: ")
print(np.round(estado, 4))
print(f"Iteraciones: {i+1}")
```

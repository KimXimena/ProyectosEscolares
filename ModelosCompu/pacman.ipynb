```python
import pygame
import sys
```

```python
TILE_SIZE = 24
MAZE = [
    "11111111111111111111",
    "1P....1...........11",
    "1111.11111.11111..11",
    "1001...........1..11",
    "1..1111.11111.11.111",
    "1..................1",
    "11111111111111111111",
]

ROWS = len(MAZE)
COLS = len(MAZE[0])
WIDTH = COLS * TILE_SIZE
HEIGHT = ROWS * TILE_SIZE

pygame.init()
screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Mini Pacman")
clock = pygame.time.Clock()
font = pygame.font.SysFont(None, 28)

# Colores
BLACK = (0, 0, 0)
BLUE = (30, 30, 200)
YELLOW = (255, 255, 0)
WHITE = (255, 255, 255)
```

```python
walls = []
pellets = set()
player_pos = [0, 0]  # (columna, fila)

for r, row in enumerate(MAZE):
    for c, ch in enumerate(row):
        if ch == "1":
            walls.append(pygame.Rect(c * TILE_SIZE, r * TILE_SIZE, TILE_SIZE, TILE_SIZE))
        elif ch == ".":
            pellets.add((c, r))
        elif ch == "P":
            player_pos = [c, r]

score = 0

def can_move(nx, ny):
    """Verifica si el jugador puede moverse a la celda (nx, ny) sin chocar con muros."""
    player_rect = pygame.Rect(nx * TILE_SIZE + 4, ny * TILE_SIZE + 4,
                              TILE_SIZE - 8, TILE_SIZE - 8)
    for w in walls:
        if player_rect.colliderect(w):
            return False
    return True
```

```python
running = True
while running:
    clock.tick(10)  # FPS bajitos para que no vaya tan rápido

    # Eventos
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    # Movimiento por teclas
    keys = pygame.key.get_pressed()
    dx = dy = 0
    if keys[pygame.K_LEFT]:
        dx = -1
    elif keys[pygame.K_RIGHT]:
        dx = 1
    elif keys[pygame.K_UP]:
        dy = -1
    elif keys[pygame.K_DOWN]:
        dy = 1

    # Intentar mover una celda
    if dx != 0 or dy != 0:
        nx = player_pos[0] + dx
        ny = player_pos[1] + dy
        if 0 <= nx < COLS and 0 <= ny < ROWS and can_move(nx, ny):
            player_pos[0] = nx
            player_pos[1] = ny

    # Comer pellet
    if (player_pos[0], player_pos[1]) in pellets:
        pellets.remove((player_pos[0], player_pos[1]))
        score += 10
        screen.fill(BLACK)

    # Muros
    for w in walls:
        pygame.draw.rect(screen, BLUE, w)

    # Pellets
    for (c, r) in pellets:
        x = c * TILE_SIZE + TILE_SIZE // 2
        y = r * TILE_SIZE + TILE_SIZE // 2
        pygame.draw.circle(screen, WHITE, (x, y), 4)

    # Jugador (Pacman)
    px = player_pos[0] * TILE_SIZE + TILE_SIZE // 2
    py = player_pos[1] * TILE_SIZE + TILE_SIZE // 2
    pygame.draw.circle(screen, YELLOW, (px, py), TILE_SIZE // 2 - 2)

    # Puntuación
    text = font.render(f"Puntos: {score}", True, WHITE)
    screen.blit(text, (10, 5))

    # Mensaje de victoria
    if not pellets:
        win_text = font.render("¡Ganaste! :)", True, YELLOW)
        screen.blit(win_text, (WIDTH // 2 - win_text.get_width() // 2,
                               HEIGHT // 2))

    pygame.display.flip()

pygame.quit()
sys.exit()
```

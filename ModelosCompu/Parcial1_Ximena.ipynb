```python
import numpy as np
import matplotlib.pyplot as plt
import os
import pandas as pd

import warnings
from IPython.display import Image, display
warnings.filterwarnings('ignore')
import seaborn as sns
from sklearn import linear_model

import openpyxl
from sklearn.metrics import r2_score
```

```python
candy_df_filepath = os.path.abspath("Regresion_Lineal/candy.csv")
df = pd.read_csv("candy.csv")
df
```

```python
x = df[['pricepercent', 'winpercent']]
y = df[['sugarpercent']]
```

```python
regr = linear_model.LinearRegression()
regr.fit(x, y)
```

```python
#Coeficiente de correlacion
print('Coeficientes de correlacion: ', regr.coef_)
```

```python
#Interseccion
print('Interseccion: ', regr.intercept_)
```

```python
#Predecir las emisiones de precios de un dulce cuyo porcentaje de precio es de 0.860 y su ganacia es de 67.60
predicted_precio = regr.predict([[0.860, 0.116], [67.60, 39.01]])
print('Emisiones de precios predichos', predicted_precio)
```

```python
#coefieneite de determinacion
modelo=linear_model
r2=r2_score(df[y],modelo.predict(df[[x]]))
print("coeficinete de deteminacion:" ,round(r2,3))
```

```python

plt.figure()
plt.scatter(x= 'pricepercent''winpercent', y = 'sugarpercent')
plt.title("Porcentaje de precio de un producto")
plt.xlabel("Precio")
plt.ylabel("Porcentaje")
plt.grid(True, linestyle="--", linewidth=0.5)
plt.show()
```

```python

plt.figure()
plt.plot(x, y, label="Ajuste lineal")
plt.title("Regresión lineal simple: Peso vs Rendimiento")
plt.xlabel("Peso (kg)")
plt.ylabel("Rendimiento (km/L)")
plt.legend()
plt.grid(True, linestyle="--", linewidth=0.5)
plt.show()
```

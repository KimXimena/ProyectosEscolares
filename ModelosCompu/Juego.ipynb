```python
import random
import os
```

```python

PALABRAS = [
    "PYTHON",
    "GALLETAS",
    "COOKIE",
    "SISTEMAS",
    "AHORCADO",
    "TRABAJO",
]

intentos_permitidos = 6  # <- número, no set
```

```python
AHORCADO_DIBUJOS = [
    """
     +---+
     |   |
         |
         |
         |
         |
    =========
    """,
    """
     +---+
     |   |
     O   |
         |
         |
         |
    =========
    """,
    """
     +---+
     |   |
     O   |
     |   |
         |
         |
    =========
    """,
    """
     +---+
     |   |
     O   |
    /|   |
         |
         |
    =========
    """,
    """
     +---+
     |   |
     O   |
    /|\\  |
         |
         |
    =========
    """,
    """
     +---+
     |   |
     O   |
    /|\\  |
    /    |
         |
    =========
    """,
    """
     +---+
     |   |
     O   |
    /|\\  |
    / \\  |
         |
    =========
    """
]
```

```python
def limpiar_pantalla():
    os.system('cls' if os.name == 'nt' else 'clear')

def elegir_palabra():
    return random.choice(PALABRAS)
```

```python
def mostrar_tablero(palabra_secreta, letras_correctas, letras_incorrectas, intentos_restantes):
    limpiar_pantalla()
    # aquí ambos son int ✅
    intentos_usados = intentos_permitidos - intentos_restantes
    print(AHORCADO_DIBUJOS[intentos_usados])

    # Mostrar la palabra
    mostrada = []
    for letra in palabra_secreta:
        if letra in letras_correctas:
            mostrada.append(letra)
        else:
            mostrada.append("_")
    print("Palabra:", " ".join(mostrada))
    print()

    if letras_incorrectas:
        print("Letras incorrectas:", ", ".join(sorted(letras_incorrectas)))
    print(f"Intentos restantes: {intentos_restantes}")
    print("-" * 30)
```

```python
def pedir_letra(letras_usadas):
    while True:
        letra = input("Ingresa una letra: ").strip().upper()
        if len(letra) != 1 or not letra.isalpha():
            print("Ingresa SOLO una letra (A-Z).")
            continue
        if letra in letras_usadas:
            print("Ya usaste esa letra, prueba con otra.")
            continue
        return letra
```

```python

def jugar_ahorcado():
    palabra = elegir_palabra()
    letras_correctas = set()
    letras_incorrectas = set()
    # Aquí nos aseguramos de que sea un número
    intentos_restantes = intentos_permitidos

    while True:
        mostrar_tablero(palabra, letras_correctas, letras_incorrectas, intentos_restantes)

        # Ganar
        if all(letra in letras_correctas for letra in palabra):
            print(f"¡Felicidades! Adivinaste la palabra: {palabra}")
            break

        # Perder
        if intentos_restantes == 0:
            print(f"Te quedaste sin intentos. La palabra era: {palabra}")
            break

        letra = pedir_letra(letras_correctas.union(letras_incorrectas))

        if letra in palabra:
            letras_correctas.add(letra)
            print(f"La letra '{letra}' SÍ está en la palabra.")
        else:
            letras_incorrectas.add(letra)
            intentos_restantes -= 1  # <- aquí también tratamos a intentos_restantes como número
            print(f"La letra '{letra}' NO está. Pierdes un intento.")

        input("\nPresiona ENTER para continuar...")

if __name__ == "__main__":
    while True:
        jugar_ahorcado()
        otra = input("\n¿Quieres jugar otra vez? (s/n): ").strip().lower()
        if otra != "s":
            print("Bye, gracias por jugar ")
            break
```

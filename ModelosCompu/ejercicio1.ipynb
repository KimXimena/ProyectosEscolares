```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
nombres = ["Jose", "Juan", "Karla", "Sandra"]

calificaciones = [[5, 5.5 , 10], [9, 8, 7, 10], [3, 5, 7], [10, 10, 10]]

df = pd.DataFrame({'Nombre': nombres , 'Calificaciones' : calificaciones})

df

medias = []
for x in df['Calificaciones']:
    medias.append(sum(x)/len(x))

df['Media'] = medias

cnt = 0

plt.bar(df['Nombre'] , df['Media'] , color = 'darkorchid')
plt.xlabel('Estudiantes')
plt.ylabel('Promedios')
plt.title('Promedios de alumnos')
for i in df['Media']:
        plt.text(cnt, i, f'{i:.4F}', ha = 'center', va = 'bottom')
        cnt +=1
plt.show()
```

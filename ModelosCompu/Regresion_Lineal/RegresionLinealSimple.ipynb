```python
# --- Regresión lineal simple: Peso (kg) -> Rendimiento (km/L) ---
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Datos
X = np.array([800, 900, 1000, 1100, 1200, 1300, 1400], dtype=float)
Y = np.array([19, 18, 17, 16, 15, 14, 13], dtype=float)
```

```python
# Añadir columna de 1s para el intercepto: modelo Y = b0 + b1*X
A = np.vstack([np.ones_like(X), X]).T
b0, b1 = np.linalg.lstsq(A, Y, rcond=None)[0]

print(f"Recta de regresión: Y = {b0:.4f} + {b1:.6f} * X")
```

```python
# Interpretación de b1 por 100 kg
print(f"Si el peso aumenta 100 kg, el rendimiento cambia {100*b1:.4f} km/L.")
```

```python
# Predicción para 1250 kg
x_pred = 1250.0
y_pred = b0 + b1 * x_pred
print(f"Predicción para 1250 kg: {y_pred:.4f} km/L")
```

```python
# Gráfico con línea ajustada
x_line = np.linspace(X.min()-50, X.max()+50, 100)
y_line = b0 + b1 * x_line

plt.figure()
plt.scatter(X, Y, label="Datos")
plt.plot(x_line, y_line, label="Ajuste lineal")
plt.title("Regresión lineal simple: Peso vs Rendimiento")
plt.xlabel("Peso (kg)")
plt.ylabel("Rendimiento (km/L)")
plt.legend()
plt.grid(True, linestyle="--", linewidth=0.5)
plt.show()
```

```python
# R^2
y_hat = b0 + b1 * X
ss_res = np.sum((Y - y_hat)**2)
ss_tot = np.sum((Y - Y.mean())**2)
r2 = 1 - ss_res/ss_tot
print(f"R^2 = {r2:.4f}")
```

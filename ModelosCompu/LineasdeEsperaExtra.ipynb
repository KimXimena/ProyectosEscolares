```python
import math
```

```python
# Parámetros del sistema
lamb = 9      # tasa de llegadas (λ)
mu = 12       # tasa de servicio por servidor (μ)
s = 3         # número de servidores (médicos)

# Cálculo de la utilización del sistema (rho)
rho = lamb / (s * mu)
rho
```

```python
# Cálculo de P0
sumatoria = 0
for n in range(s):
    sumatoria += (lamb/mu)**n / math.factorial(n)

# Segunda parte de la fórmula
parte2 = ((lamb/mu)**s) / math.factorial(s) * (1 / (1 - rho))

P0 = 1 / (sumatoria + parte2)
P0
```

```python
Lq = (((lamb/mu)**s) * rho * P0) / (math.factorial(s) * (1 - rho)**2)
Lq
```

```python
L = Lq + (lamb/mu)
L
```

```python
Wq = Lq / lamb
Wq
```

```python
W = Wq + (1/mu)
W
```

```python
print("RESULTADOS DEL MODELO M/M/3 EN EL ÁREA DE EMERGENCIAS\n")

print(f"λ (llegadas por hora): {lamb}")
print(f"μ (servicio por hora por médico): {mu}")
print(f"Servidores S: {s}")
print(f"Utilización ρ: {rho:.4f}\n")

print(f"P0 (probabilidad de cero pacientes): {P0:.4f}")
print(f"Lq (pacientes en cola): {Lq:.4f}")
print(f"L  (pacientes en el sistema): {L:.4f}")
print(f"Wq (tiempo en cola en horas): {Wq:.4f} h  ({Wq*60:.2f} minutos)")
print(f"W  (tiempo total en el sistema): {W:.4f} h  ({W*60:.2f} minutos)")
```

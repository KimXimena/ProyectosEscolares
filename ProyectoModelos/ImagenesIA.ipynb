```python
import cv2
import numpy as np

# Cargar modelo
modelo = load_model("numeros_conv_ad_do.h5")

# Cargar una imagen de prueba
img = cv2.imread("test.png", cv2.IMREAD_GRAYSCALE)
img = cv2.resize(img, (28, 28))       # <- cambia si tu modelo usa otro tamaño
img = img.astype("float32") / 255.0
img = img.reshape(1, 28, 28, 1)       # <- si es blanco y negro

# Predicción
pred = modelo.predict(img)
clase = np.argmax(pred)

print("Predicción:", clase)
```

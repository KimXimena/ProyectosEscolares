```python
# Entrenamiento MNIST CNN 
import tensorflow as tf
from keras.datasets import mnist
from keras.utils import to_categorical
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import numpy as np
import matplotlib.pyplot as plt
```

```python
#carga del dataset MNIST
(X_entrenamiento, Y_entrenamiento), (X_pruebas, Y_pruebas) = mnist.load_data()
```

```python
#cambio de imágenes para que sean blanco y negro y las normalizo
#entre 255, porque las CNN aprenden mejor con valores entre 0 y 1
X_entrenamiento = X_entrenamiento.reshape(-1,28,28,1).astype('float32')/255.
X_pruebas = X_pruebas.reshape(-1,28,28,1).astype('float32')/255.
```

```python
#convierto las etiquetas a formato one-hot para que la salida use 10 neuronas
Y_entrenamiento = to_categorical(Y_entrenamiento)
Y_pruebas = to_categorical(Y_pruebas)
```

```python
#aumnto de imagenes, rotaciones, desplazaminetos y zoom
datagen = ImageDataGenerator(rotation_range=30,width_shift_range=0.25,
                                 height_shift_range=0.25, zoom_range=[0.5,1.5])
datagen.fit(X_entrenamiento) #ajuste del genarador con mis imagenes
```

```python
#modelo CNN
modelo = tf.keras.models.Sequential([
    tf.keras.layers.Conv2D(32,(3,3),activation='relu',input_shape=(28,28,1)),
    tf.keras.layers.MaxPooling2D(2,2),
    tf.keras.layers.Conv2D(64,(3,3),activation='relu'),
    tf.keras.layers.MaxPooling2D(2,2),
    tf.keras.layers.Dropout(0.5),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(100,activation='relu'),
    tf.keras.layers.Dense(10,activation='softmax')])
modelo.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
#lote de 32 imagenes
batch=32
data_gen = datagen.flow(X_entrenamiento,Y_entrenamiento,batch_size=batch) #que cada lote venga con imagenes distintas
#entrenamiento de 10 epocas
history = modelo.fit(data_gen, epochs=10,
                      validation_data=(X_pruebas,Y_pruebas),
                      steps_per_epoch=int(np.ceil(60000/batch)), #cuantas imagenes procesa por epoca
                      validation_steps=int(np.ceil(10000/batch)))
```

```python
modelo.save('numeros_conv_ad_do.h5')
```

```python
from tensorflow import keras
import numpy as np
from PIL import Image
import gradio as gr

modelo = keras.models.load_model("numeros_conv_ad_do.h5")
```

```python
def predecir_digito(imagen):
    if imagen is None:
        return "No se recibió imagen"

    if isinstance(imagen, np.ndarray):
        imagen = imagen.astype("uint8")
        img = Image.fromarray(imagen)
    else:
        img = imagen

    img = img.convert("L")
    img = img.resize((28, 28))

    arr = np.array(img).astype("float32") / 255.0

    arr = 1.0 - arr     #dígito blanco y fondo negro

    arr = arr.reshape(1, 28, 28, 1)

    pred = modelo.predict(arr)
    digito = int(np.argmax(pred))

    return f"Predicción: {digito}"
```

```python
demo = gr.Interface(
    fn=predecir_digito,
    inputs=gr.Image(label="Sube una imagen de un dígito (0–9)"),
    outputs=gr.Textbox(label="Predicción"),
    title="Reconocimiento de dígitos",
    description="Sube una imagen en la que aparezca un número del 0 al 9. El modelo CNN entrenado con MNIST intentará reconocerlo."
)

demo.launch()
```

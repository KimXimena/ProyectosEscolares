```python
import tensorflow as tf
from tensorflow import keras
from keras.datasets import mnist
import numpy as np
import matplotlib.pyplot as plt
```

```python
#leo los datos 
modelo = keras.models.load_model('numeros_conv_ad_do.h5')
```

```python
#modelo one hot encoding, para medir que tan bien predicce
(_, _), (X_pruebas, Y_pruebas) = mnist.load_data()
X_pruebas = X_pruebas.reshape(-1,28,28,1).astype('float32')/255.
from keras.utils import to_categorical
Yp = to_categorical(Y_pruebas)
loss,acc = modelo.evaluate(X_pruebas,Yp,verbose=0)
print(acc)
```

```python
#prediccion de las primeras 8 imagenes
pred = modelo.predict(X_pruebas[:8])
fig,axes=plt.subplots(1,8,figsize=(16,2))
for i in range(8):
    axes[i].imshow(X_pruebas[i].reshape(28,28),cmap='gray_r')
    axes[i].set_title(np.argmax(pred[i]))
    axes[i].axis('off')
plt.show()
```

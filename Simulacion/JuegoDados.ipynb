```python
import random
import matplotlib.pyplot as plt
import numpy as np
```

```python
interacciones = 25
par1 = 0
par2 = 0
par3 = 0
par4 = 0
par5 = 0
par6 = 0
impar = 0

gana_par1 = (1*1) * 100
gana_par2 = (1*2) * 100
gana_par3 = (1*3) * 100
gana_par4 = (1*4) * 100
gana_par5 = (1*5) * 100
gana_par6 = (1*6) * 100
gana_impar = -200

for i in range(interacciones):
    dado1 = random.randint(1, 6)
    dado2 = random.randint(1, 6)
    #print(f"Tirada {i+1}: Dado 1: {dado1}, Dado 2: {dado2}")

    if (dado1==dado2 and dado1 == 1):
        par1 += 1
    elif (dado1==dado2 and dado1 == 2):
        par2 += 1
    elif (dado1==dado2 and dado1 == 3):
        par3 += 1
    elif (dado1==dado2 and dado1 == 4):
        par4 += 1
    elif (dado1==dado2 and dado1 == 5):
        par5 += 1
    elif (dado1==dado2 and dado1 == 6):
        par6 += 1
    else:
        impar += 1
        
resultados=np.array([impar,par1,par2,par3,par4,par5,par6,])
ganancias=np.array([gana_impar,gana_par1,gana_par2,gana_par3,gana_par4,gana_par5,gana_par6,])
print("Resultados de las tiradas:",resultados)
print("Ganancias por cada tipo de tirada:",ganancias)
```

```python
#graficar
plt.title("Resultados del juego de dados")
plt.ylabel("Cantidad de Tiros")
plt.bar(np.arange(len(ganancias)),resultados,tick_label=['Impar','Par 1','Par 2','Par 3','Par 4','Par 5','Par 6'])
cnt=0
for ganancia in ganancias:
    plt.text(cnt,resultados[cnt],f"${ganancia}",ha='center',va='bottom')
    cnt+=1
plt.show()
```

```python
loquegano=0
for i in range(0,7):
    print(resultados[i]*ganancias[i])
    if (i==0):
        perdidas=resultados[i]*ganancias[i]
    else:
        loquegano+=resultados[i]*ganancias[i]
print("Ganancias totales después de", interacciones, "interacciones: $", loquegano+perdidas)
```

```python
# Calcular ganancias totales
total_ganancias = np.sum(resultados * ganancias)
print("Ganancias totales después de", interacciones, "interacciones: $", total_ganancias)
```

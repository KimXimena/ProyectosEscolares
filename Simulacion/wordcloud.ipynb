```python
from wordcloud import WordCloud
import nltk 
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np
import requests
```

```python
# cargar datos
url_libro = "https://www.gutenberg.org/cache/epub/5654/pg5654.txt"
libro = requests.get(url_libro).text
print(libro)
```

```python
nube = WordCloud().generate(libro)
plt.imshow(nube)
plt.show
```

```python
#eliminar plabras vacias
nltk.download('stopwords')
palabras_irrelevantes=nltk.corpus.stopwords.words("english")
```

```python
palabras_irrelevantes
```

```python
#agregar mas palabras irrelevantes
palabras_irrelevantes.extend(["JOHN", "DAVID, JAMES, MAGGIE"])
```

```python
#nueva nube de palabras con stopwords
nube = WordCloud(stopwords=palabras_irrelevantes).generate(libro)
plt.imshow(nube)
plt.axis("off")
plt.show()
```

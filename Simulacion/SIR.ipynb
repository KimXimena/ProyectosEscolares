```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#parametros del modelo
beta = 0.3 #tasa de trasmisión
gamma = 0.1 #tasa de recuperación
n = 1000 #población total
I0 = 1 #numer inicial de infectados
R0 = 0 #numero inicial de recuperados
S0 = n - I0 #numero de iniciales susceptibles
```

```python
#definir ecuaciones diferenciales
def sir_model(y, t, beta, gamma, n):
    S, I, R = y
    dSdt = -beta * S * I / n
    dIdt = beta * S * I / n - gamma * I
    dRdt = gamma * I
    return [dSdt, dIdt, dRdt]
```

```python
#tiempo de simulación
t = np.linspace(0, 260, 260) #160 días
```

```python
#condiciones iniciales
y0 = [S0, I0, R0]
```

```python
#resover las ecuaciones diferenciales
from scipy.integrate import odeint
#odeint dy / dt = f(y, t)
#y0: condiciones iniciales
#t: tiempo de simulación
#parametros = beta, gamma, n
solution = odeint(sir_model, y0, t, args=(beta, gamma, n))
```

```python
#extraer los resultados
S, I, R = solution.T
```

```python
S, I, R
```

```python
#graficar
plt.figure(figsize=(10, 6))
plt.plot(t, S, label='Suceptibles (S)', color='blue')
plt.plot(t, I, label='Infectados (I)', color='red')
plt.plot(t, R, label='Recuperados (R)', color='green')
plt.xlabel('Tiempo (días)')
plt.ylabel('Número de personas')
plt.title('Simulación de epidemia modelo SIR')
plt.legend()
plt.grid=True
plt.show()
```

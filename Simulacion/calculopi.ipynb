```python
import random
rango = 20000
```

```python
circle_points = 0
square_points = 0

for _ in range(rango):
    rand_x = random.uniform(-1, 1)
    rand_y = random.uniform(-1, 1)
    distance = (rand_x**2 + rand_y**2)**0.5
    if distance <= 1:
        circle_points += 1
    square_points += 1
pi_estimate = 4 * circle_points / square_points
print(f"El valor aproximado de pi es: {pi_estimate}")
```

```python
def probar(valor_casilla_1):
    casillas = [0] * 16

    casillas[1] = valor_casilla_1
    casillas[3] = 5
    casillas[6] = 7
    casillas[9] = 4
    casillas[12] = 6

    pos = 3
    duplicadas = []

    while casillas[1] >= pos:
        casillas[pos] = casillas[pos] * 2
        duplicadas.append(pos)
        pos = pos + 3

    print(f"\nValor en casilla 1: {valor_casilla_1}")
    print("Casillas duplicadas:", duplicadas)
    print("Siguiente posición donde se detuvo:", pos)

    for i in [3, 6, 9, 12]:
        print(f"Casilla {i} = {casillas[i]}")


for valor in [11, 12, 13, 14, 15]:
    probar(valor)
```
